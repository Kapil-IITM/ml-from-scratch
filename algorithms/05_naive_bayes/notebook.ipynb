# Naive Bayes from Scratch

Naive Bayes is a probability-based algorithm, built on Bayes' Theorem.
Unlike the algorithms covered so far, it doesn't learn weights, compare
distances, or split data with questions - it works by calculating
probabilities directly from the data and picking whichever class is more
likely.

## Bayes' Theorem

This whole algorithm is based on Bayes' Theorem, which is basically a
formula for updating probability when we get new evidence.

$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

In simple words - probability of A given B equals probability of B given
A, times probability of A, divided by probability of B.

For our placement data, what we actually want to know is: given a
student's features (CGPA, internship, projects), what's the probability
they got placed? So the formula becomes:

$$P(Placed|Features) = \frac{P(Features|Placed) \cdot P(Placed)}{P(Features)}$$

We calculate this same thing for both "Placed" and "Not Placed", and
just go with whichever one is bigger.

A couple of terms to keep in mind here:
- **P(Placed)** is called the prior - just how common placement is
  overall in the data, before we even look at any features
- **P(Features|Placed)** is the likelihood - how likely these specific
  features are, assuming the student was placed
- **P(Features)** ends up being the same for both classes when we
  compare them, so we can basically ignore it - it's not going to change
  which class wins anyway

# Why "Naive"?

So here is the catch with this algorithm. To actually figure out $P(Features|Placed)$, we would normally need to figure out how things like CGPA, internships, and projects all interact with each other in the real world. That gets super messy and complicated really fast.

Naive Bayes gets around this by just assuming all our features are completely independent of each other. Instead of calculating one giant, messy joint probability, we just multiply the individual probabilities together:

$$P(Features|Placed) = P(CGPA|Placed) \cdot P(Internship|Placed) \cdot P(Projects|Placed)$$

Now, we call this assumption "naive" because in real life, things are rarely independent. For instance, students with a high CGPA usually tend to do more projects, right? 

Even though this assumption is technically wrong in practice, Naive Bayes still works surprisingly well and gets the job done. That is pretty much why we keep using it!

## Handling Continuous Features (Gaussian Naive Bayes)

There's a problem though - P(CGPA|Placed) doesn't really make sense the
way we've been thinking about probability so far, since CGPA can be any
decimal value (like 7.34 or 8.921), not a fixed category like
"placed/not placed".

This is where **Gaussian Naive Bayes** comes in. It assumes each
continuous feature follows a normal (bell curve) distribution within
each class, and uses the normal distribution formula to estimate the
probability:

$$P(x|class) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

Here, $\mu$ and $\sigma^2$ are just the mean and variance of that feature
within that class. So for CGPA specifically, we'd calculate the mean and
variance of CGPA separately for placed students and for not-placed
students, and plug those into this formula whenever we need to estimate
a probability.

For features that are just 0/1 (like internship) or small counts (like
number of projects), we could technically still use this same Gaussian
formula, even though it's not a perfect fit - it's common to just treat
everything as Gaussian for simplicity, which is what we'll do here.
